```python
catalog = spark.sql(
    "SELECT current_catalog()"
).first()[0]

spark.sql(f"""
    SHOW TABLES IN `{catalog}`.`retail_lakehouse`
""").show(truncate=False)
```

```python
spark.sql(f"""
    SELECT check_name, status, failed_records, run_timestamp
    FROM `{catalog}`.`retail_lakehouse`.`data_quality_audit`
    ORDER BY run_timestamp DESC, check_name
""").show(truncate=False)
```
